```python
import tensorflow as tf
keras = tf.keras
from keras.layers import MultiHeadAttention

# Example input: batch_size=2, sequence_length=5,
# embedding_dimension=16
inputs = tf.random.normal((2, 5, 16))
```

```python
attention = MultiHeadAttention(
    num_heads=4,
    key_dim=4
)

outputs = attention(
    query=inputs,
    key=inputs,
    value=inputs
)

print(outputs.shape)
```
