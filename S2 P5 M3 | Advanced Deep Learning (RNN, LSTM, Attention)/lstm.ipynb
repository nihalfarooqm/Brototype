```python
import tensorflow as tf
keras = tf.keras
from keras.layers import Input, Dense, LSTM
from keras import Model
```

```python
inputs = Input(shape=(10, 4))

x = LSTM(
    units=32,
    return_sequences=False
)(inputs)

outputs = Dense(
    1,
    activation="sigmoid"
)(x)

model = Model(inputs, outputs)

model.summary()
```
